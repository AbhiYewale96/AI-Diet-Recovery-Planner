**SECTION 2: DAILY TRACKING DATASET PROCESSING**

**1. Load Daily Tracking Dataset**

```python
df1 = pd.read_csv("/content/daily_tracking_expanded (1).csv")
```

**2. Checking Missing Values**

```python
df1.isnull()
```

```python
df1.isnull().sum()
```

**3. Data Preprocessing and Feature Extraction for Daily Tracking**

**4. Handling & Verify Missing Values**

```python
import numpy as np

# Clean column names
df1.columns = df1.columns.str.strip().str.lower()

# Handle categorical columns
categorical_cols = ['exercise_done', 'symptoms', 'notes']

for col in categorical_cols:
    if col in df1.columns:
        df1[col] = df1[col].fillna('None')

# Handle blood pressure
if 'blood_pressure' in df1.columns:

    # Convert to string
    bp_split = df1['blood_pressure'].astype(str).str.split('/', expand=True)

    # Create systolic and diastolic columns
    df1['systolic_bp'] = pd.to_numeric(bp_split[0], errors='coerce')
    df1['diastolic_bp'] = pd.to_numeric(bp_split[1], errors='coerce')

    # Fill missing values with median
    df1['systolic_bp'] = df1['systolic_bp'].fillna(df1['systolic_bp'].median())
    df1['diastolic_bp'] = df1['diastolic_bp'].fillna(df1['diastolic_bp'].median())

    # Drop original column
    df1.drop(columns=['blood_pressure'], inplace=True)

# Numerical columns
numerical_cols = ['weight_kg', 'calorie_intake', 'protein_intake', 'carb_intake', 'fat_intake',
                  'steps_count', 'sleep_hours', 'blood_sugar_level', 'heart_rate']

# Convert columns to numeric safely
for col in numerical_cols:
    if col in df1.columns:
        df1[col] = pd.to_numeric(df1[col], errors='coerce')

        # Fill missing values with median
        df1[col] = df1[col].fillna(df1[col].median())

# Final null check
print("Remaining Null Values:")
print(df1.isnull().sum())

print("\nDataset Info:")
print(df1.info())

print("\nFirst 5 Rows:")
display(df1.head())
```

**5. Validate datatypes**

```python
df1['date'] = pd.to_datetime(df1['date'], format = '%d-%m-%Y')
print("\n----- Data types ['date'] after date\n")
df1.info()
df1.head()
```

**6. Standardise categorical values**

```python
categorical_col = df1.select_dtypes(include=['object']).columns
Standardise_col = ['record_id','patient_id', 'meal_compliance', 'exercise_done',
       'symptoms', 'energy_level', 'mood_status', 'medication_taken',
       'bowel_movement_status', 'blood_pressure', 'doctor_visit',
       'notes']

for col in Standardise_col:
    if col in df1.columns:
        df1[col] = df1[col].astype(str).str.title().str.strip()

print("Data types after categorical standardization:")
df1.info()
df1.head()
```

**7. Validate realistic data ranges**

```python
print("\n------- Numerical data range validation -----\n")

# General check for all numerical columns
numerical_cols = df1.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_cols:
    print(f"Column: {col}")
    print(f"  Min: {df1[col].min()}")
    print(f"  Max: {df1[col].max()}")
    print(f"  Mean: {df1[col].mean():.2f}")
    print(f"  Median: {df1[col].median():.2f}")
    print("--------------------------------")

# Validate realistic data ranges
# Weight in kg: 2 - 120
df1 = df1[df1['weight_kg'].between(2, 300)]

# Calorie intake: 0 - 10000
df1 = df1[df1['calorie_intake'].between(0, 10000)]

# Protein intake: 0 - 500 grams
df1 = df1[df1['protein_intake'].between(0, 500)]

# Carbohydrate intake: 0 - 1000 grams
df1 = df1[df1['carb_intake'].between(0, 1000)]

# Fat intake: 0 - 500 grams
df1 = df1[df1['fat_intake'].between(0, 500)]

# Steps count: 0 - 100000
df1 = df1[df1['steps_count'].between(0, 100000)]

# Sleep hours: 0 - 24
df1 = df1[df1['sleep_hours'].between(0, 24)]

# Blood sugar level: 40 - 500 mg/dL
df1 = df1[df1['blood_sugar_level'].between(40, 500)]

# Heart rate: 30 - 220 bpm
df1 = df1[df1['heart_rate'].between(30, 220)]

# Systolic BP: 70 - 250
df1 = df1[df1['systolic_bp'].between(70, 250)]

# Diastolic BP: 40 - 150
df1 = df1[df1['diastolic_bp'].between(40, 150)]

# -----------------------------------
# Final dataset shape
# -----------------------------------

print("Dataset shape after validation:")
print(df1.shape)

print("\nRemaining Null Values:")
print(df1.isnull().sum())

```

**8. Save & Download the dataset**
